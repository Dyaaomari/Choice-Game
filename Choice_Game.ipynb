```python
print('Welcome to CHOSEN!')

name = input('What is your name? ')
age = int(input('What is your age? '))

health = 20
print(f'You have {health} healt points. So make your choses carefully!')

# SCENARIO (1.0) - {Age is greater than or equal to 18}
if age >= 18:
    print("You are old enough to play!")

    wants_to_play = input('Do you want to play? ').lower()
    pick_a_weapon = input('Pick the right weapon to protect yourself! (sword/gun)? ').lower()

    # 1.1 - {The gamer choses yes and sword}
    if wants_to_play == 'yes' and pick_a_weapon == 'sword':
        print("Let's play!")
        left_or_right = input('First choice... Left way or Right way (left/right)? ').lower()

        # 1.1.1 - {The gamer choses the left way}
        if left_or_right == 'left':
            ans = input('Nice, you follow the path and reach a lake... Do you swim across or go around (across/around)? ').lower()

            # 1.1.1.1 - {The gamer choses across}
            if ans == 'across':
                print('You managed to get across but were bit by a fish and lost 15 health, but since you have your sword, you only lose 5 health.')
                health -= 5

            # 1.1.1.2 - {The gamer choses around}
            elif ans == 'around':
                print('You went around and reached the other side of the lake')

            ans = input('You notice a house and a river. Which do you go to (river/house)? ').lower()

            # 1.1.1.2.1 - {The gamer choses house}
            if ans == 'house':
                print("You go to the house and are caught by the owner... He doesn't like it, but he scurries from your sword, so you nothing")
                health -= 5
                if health <= 0:
                    print('You now have 0 health and lose the game...')
                else:
                    print('You have survived...You win!')

            # 1.1.1.2.2 - {The gamer choses river}
            else:
                print('You fell in the river and lost...')

        # 1.1.2 - {The gamer choses the right way and lost!}
        else:
            print('You lost! You took the right way.')

    # 1.2 - {The gamer choses yes and gun}
    elif wants_to_play == 'yes' and pick_a_weapon == 'gun':
        print("Let's play with firepower!")
        left_or_right = input('First choice... Left way or Right way (left/right)? ').lower()

        # 1.2.1 - {The gamer choses the left way}
        if left_or_right == 'left':
            ans = input('Nice, you follow the path and reach a lake... Do you swim across or go around (across/around)? ').lower()

            # 1.2.1.1 - {The gamer choses across}
            if ans == 'across':
                print('You swam across, but your gun got wet and ruined! A giant fish bit you and you lost 10 health.')
                health -= 10

            # 1.2.1.2 - {The gamer choses around}
            elif ans == 'around':
                print('You walked around the lake safely and kept your gun dry.')

            ans = input('You notice a house and a river. Which do you go to (river/house)? ').lower()

            # 1.2.1.2.1 - {The gamer choses house}
            if ans == 'house':
                print("You enter the house. The owner sees your gun, gets scared, and surrenders!")
                if health <= 0:
                    print('You now have 0 health and lose the game...')
                else:
                    print('You have survived... You win!')

            # 1.2.1.2.2 - {The gamer choses river}
            else:
                print('You slipped into the river under the weight of your weapon and lost...')

        # 1.2.2 - {The gamer choses the right way and lost!}
        else:
            print('You took the right way and got ambushed... You lost!')

    # 1.3 - {The gamer choses no or wrong input}
    else:
        print('Bye...')

# SCENARIO (2.0) - {Age is younger than 18}
else:
    print("You are not old enough to play...")
```
